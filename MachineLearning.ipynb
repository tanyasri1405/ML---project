```python
plt.figure(figsize=(6,4))

sns.countplot(x=" Label", data=df)

plt.title("Legitimate vs Fraudulent Transactions")
plt.xlabel("Class")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
class_counts = df[" Label"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    class_counts,
    labels=class_counts.index.tolist(), # Use actual labels from value_counts()
    autopct="%1.2f%%"
)

plt.title("Transaction Class Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Flow Bytes/s"], bins=50)

plt.title("Distribution of Flow Bytes per Second")
plt.xlabel("Flow Bytes/s")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=" Label", y="Flow Bytes/s", data=df)

plt.title("Flow Bytes per Second by Label")
plt.xlabel("Label")
plt.ylabel("Flow Bytes/s")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x=" Label",
    y="Flow Bytes/s",
    data=df,
    showfliers=False
)

plt.title("Flow Bytes per Second by Label (No Outliers)")
plt.xlabel("Label")
plt.ylabel("Flow Bytes/s")

plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df[" Flow Duration"], bins=100)

plt.title("Flow Duration Distribution")
plt.xlabel("Flow Duration (microseconds)")
plt.ylabel("Number of Flows")

plt.show()
```

```python
benign_df = df[df[" Label"] == 'BENIGN']

plt.figure(figsize=(10,5))

plt.scatter(
    benign_df[" Flow Duration"],
    benign_df["Flow Bytes/s"],
    alpha=0.5
)

plt.title("Benign Flows: Flow Bytes/s vs. Flow Duration")
plt.xlabel("Flow Duration (microseconds)")
plt.ylabel("Flow Bytes/s")

plt.show()
```

```python
filtered_df = df[(df[" Label"] == 'BENIGN') & (df[" Flow Duration"] > 0) & (df["Flow Bytes/s"] != np.inf) & (df["Flow Bytes/s"] != -np.inf) & (df["Flow Bytes/s"].notna())]

print("Average Flow Bytes/s for BENIGN flows (excluding zero duration/inf values):",
      filtered_df["Flow Bytes/s"].mean())

print("Median Flow Bytes/s for BENIGN flows (excluding zero duration/inf values):",
      filtered_df["Flow Bytes/s"].median())
```

```python
plt.figure(figsize=(16,12))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    cmap="coolwarm",
    center=0,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
features = [" Flow Duration", "Flow Bytes/s", " Total Fwd Packets", " Total Backward Packets"]

for feature in features:

    plt.figure(figsize=(8,4))

    sns.kdeplot(
        data=df,
        x=feature,
        hue=" Label",
        fill=True
    )

    plt.title(f"{feature} Distribution by Label")

    plt.show()
```

```python
Q1 = df["Flow Bytes/s"].quantile(0.25)
Q3 = df["Flow Bytes/s"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Flow Bytes/s"] < lower_bound) |
        (df["Flow Bytes/s"] > upper_bound)
        ]

print("Number of Flow Bytes/s outliers:", len(outliers))
```

```python
print("========== CREDIT CARD FRAUD DETECTION - EDA SUMMARY ==========")

print("Total transactions:", len(df))

print("Total columns:", df.shape[1])

print("\nClass Distribution:")
print(df[" Label"].value_counts())

print("\nMissing Values:",
      df.isnull().sum().sum())

print("\nDuplicate Rows:",
      df.duplicated().sum())

non_benign_count = len(df[df[" Label"] != 'BENIGN'])
fraud_percentage = (non_benign_count / len(df)) * 100

print("\nNon-BENIGN Percentage:",
      round(fraud_percentage, 4), "%")

print("\nAverage Flow Bytes/s:",
      round(df["Flow Bytes/s"].mean(), 2))

print("\nMedian Flow Bytes/s:",
      round(df["Flow Bytes/s"].median(), 2))

print("\n========== EDA COMPLETED ==========")
```
